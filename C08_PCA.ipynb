# Análisis de Componentes Principales (PCA)

**Observación importante 1.**

Supongamos que tenemos una matriz cuadrada $\mathbf{A}$ con $n$ filas y $p$ columnas. Sea $\mathbf{v}$ un vector en $\mathbb{R}^p$ con $\|\mathbf{v}\|=1$.

Supongamos que las filas de $\mathbf{A}$ son $\mathbf{a}^{(1)},\mathbf{a}^{(2)},\dots,\mathbf{a}^{(n)}$. Estas filas son elementos de $\mathbb{R}^p$. Si queremos proyectar el vector $\mathbf{a}^{(k)}$ sobre el espacio generado por $\mathbf{v}$ tenemos:

$$P_{\langle \mathbf{v}\rangle}\mathbf{a}^{(k)} = \frac{\langle \mathbf{a}^{(k)},\mathbf{v}\rangle}{\|\mathbf{v}\|^2} \mathbf{v} = \langle \mathbf{a}^{(k)},\mathbf{v}\rangle \mathbf{v}$$

Por lo tanto, hacer esto con los $n$ renglones de $\mathbf{A}$ equivale a calcular $\mathbf{A}\mathbf{v}$:

$$\mathbf{A}\mathbf{v} = \begin{pmatrix} \langle \mathbf{a}^{(1)},\mathbf{v}\rangle \\ \langle \mathbf{a}^{(2)},\mathbf{v}\rangle \\ \vdots \\ \langle \mathbf{a}^{(n)},\mathbf{v}\rangle \end{pmatrix}$$

---

**Observación importante 2.**

Ahora, sea $\mathbf{B}$ una matriz con $n$ filas y $p$ columnas, donde **cada columna tiene media 0**.

Esto significa que si $B_i$ es la $i$-ésima columna de $\mathbf{B}$, entonces:

$$\frac{B_{1i} + B_{2i} + \dots + B_{ni}}{n} = 0$$

Si tomamos $\mathbf{v} \in \mathbb{R}^p$, entonces $\mathbf{B}\mathbf{v}$ tiene sentido y además:

$$\mathbf{B}\mathbf{v} = \begin{pmatrix} \langle \mathbf{b}^{(1)},\mathbf{v}\rangle \\ \langle \mathbf{b}^{(2)},\mathbf{v}\rangle \\ \vdots \\ \langle \mathbf{b}^{(n)},\mathbf{v}\rangle \end{pmatrix}$$

donde $\mathbf{b}^{(k)}$ es la $k$-ésima **fila** de $\mathbf{B}$. Entonces:

$$\overline{\mathbf{B}\mathbf{v}} = \frac{\langle \mathbf{b}^{(1)},\mathbf{v}\rangle + \dots + \langle \mathbf{b}^{(n)},\mathbf{v}\rangle}{n}
= \left\langle \frac{\mathbf{b}^{(1)}+\dots+\mathbf{b}^{(n)}}{n}, \mathbf{v} \right\rangle = 0$$

---

El **Análisis de Componentes Principales** (PCA) es una técnica de **reducción de dimensionalidad** ampliamente utilizada en estadística multivariante y aprendizaje automático.

El PCA transforma un conjunto de variables posiblemente correlacionadas en un nuevo conjunto de variables llamadas **componentes principales**. Estas nuevas variables:

- Son combinaciones lineales de las originales.
- Son incorreladas entre sí (ortogonales).
- Están ordenadas de manera que la primera explica la **mayor parte de la varianza** total de los datos, la segunda la siguiente mayor cantidad (perpendicular a la primera), y así sucesivamente.

**Objetivos principales del PCA:**
- Reducir la dimensionalidad sin perder demasiada información.
- Eliminar multicolinealidad.
- Facilitar la visualización de datos de alta dimensionalidad.
- Servir como preprocesamiento para otros algoritmos de Machine Learning.

## Preparación de los datos y matriz de correlación

Sea $\mathbf{X}$ la matriz de datos de tamaño $n \times p$, donde $n$ es el número de observaciones y $p$ el número de variables.

### 1. Estandarización de los datos

Primero centramos y escalamos cada variable para obtener la matriz estandarizada $\mathbf{Z}$:

$$\mathbf{Z} = (\mathbf{X} - \mathbf{1}\bar{\mathbf{x}}^T) \mathbf{D}^{-1}$$

donde:
- $\mathbf{1}$ es un vector columna de unos de tamaño $n \times 1$,
- $\bar{\mathbf{x}}$ es el vector de medias de tamaño $p \times 1$,
- $\mathbf{D} = \text{diag}(s_1, s_2, \dots, s_p)$ es la matriz diagonal que contiene las desviaciones estándar de cada variable.

De forma equivalente, elemento a elemento:

$$Z_{ij} = \frac{X_{ij} - \bar{X}_j}{s_j}, \quad i=1,\dots,n; \quad j=1,\dots,p$$

Después de esta transformación, cada columna de $\mathbf{Z}$ tiene **media cero** y **desviación estándar uno**.

### 2. Matriz de correlación $\mathbf{R}$

En PCA **siempre usamos la matriz de correlación** $\mathbf{R}$, que se calcula como:

$$\mathbf{R} = \frac{1}{n-1} \mathbf{Z}^T \mathbf{Z}$$

**Propiedades importantes de $\mathbf{R}$:**
- Es **simétrica**: $\mathbf{R} = \mathbf{R}^T$
- Es **semidefinida positiva**: todos sus eigenvalores satisfacen $\lambda_i \ge 0$
- Los elementos en la diagonal son iguales a 1: $R_{jj} = 1$
- Los elementos fuera de la diagonal $R_{jk}$ representan la correlación lineal entre las variables $j$ y $k$

En efecto: todas las columnas de $\mathbf{Z}$ tienen media 0 y varianza 1. Por lo tanto, si $\mathbf{x}$ e $\mathbf{y}$ son columnas de $\mathbf{Z}$, tenemos:

$$1 = \text{Var}(\mathbf{x}) = \frac{\|\mathbf{x}\|^2}{n-1}, \quad
1 = \text{Var}(\mathbf{y}) = \frac{\|\mathbf{y}\|^2}{n-1}$$

y

$$\text{Corr}(\mathbf{x},\mathbf{y}) = \frac{\langle \mathbf{x},\mathbf{y} \rangle}{\|\mathbf{x}\|\|\mathbf{y}\|} = \frac{\langle \mathbf{x},\mathbf{y} \rangle}{n-1}$$

Luego:

$$R_{jk} = \langle \mathbf{e}_j, \mathbf{R} \mathbf{e}_k \rangle
= \frac{1}{n-1} \langle \mathbf{e}_j, \mathbf{Z}^T \mathbf{Z} \mathbf{e}_k \rangle
= \frac{1}{n-1} \langle \mathbf{Z} \mathbf{e}_j, \mathbf{Z} \mathbf{e}_k \rangle$$

Como $\mathbf{Z} \mathbf{e}_j$ y $\mathbf{Z} \mathbf{e}_k$ son las columnas $j$ y $k$ de $\mathbf{Z}$, entonces:

$$\langle \mathbf{Z} \mathbf{e}_j, \mathbf{Z} \mathbf{e}_k \rangle = (n-1) \, \text{Corr}(\mathbf{Z} \mathbf{e}_j, \mathbf{Z} \mathbf{e}_k)$$

y de ahí es fácil terminar.

---

Esta matriz $\mathbf{R}$ es la que utilizaremos para realizar la descomposición espectral en el PCA.

### Eigenvalores y eigenvectores

Como $\mathbf{R}$ es simétrica y semidefinida positiva, todos sus eigenvalores son $\ge 0$ y admite la descomposición espectral:

$$\mathbf{R} = \mathbf{Q} \boldsymbol{\Lambda} \mathbf{Q}^T$$

donde:
- $\boldsymbol{\Lambda} = \text{diag}(\lambda_1, \lambda_2, \dots, \lambda_p)$ con $\lambda_1 \ge \lambda_2 \ge \dots \ge \lambda_p \ge 0$
- Las columnas de $\mathbf{Q}$ son los **eigenvectores unitarios** $\mathbf{v}_1, \mathbf{v}_2, \dots, \mathbf{v}_p$

**Componentes principales**

Por la **Observación importante 1**, proyectar los datos estandarizados $\mathbf{Z}$ (cuyas filas son observaciones) sobre un vector unitario $\mathbf{v}_i$ equivale a calcular $\mathbf{Z}\mathbf{v}_i$. Por lo tanto, la $i$-ésima componente principal se define como:

$$\mathbf{Y}_i = \mathbf{Z} \mathbf{v}_i$$

Cada $\mathbf{Y}_i$ es un vector columna de longitud $n$ que contiene las coordenadas de las $n$ observaciones proyectadas sobre la dirección $\mathbf{v}_i$.

Por lo tanto $$\mathbf{Y} = \mathbf{Z} \mathbf{Q}$$ es la matriz de componentes principales.

Como las columnas de $\mathbf{Z}$ tienen media 0, por la **Observación importante 2** tenemos:

$$\overline{\mathbf{Y}_i} = \overline{\mathbf{Z} \mathbf{v}_i} = 0$$

De lo anterior, la **varianza** de esta componente principal es:

$$\text{Var}(\mathbf{Y}_i) = \text{Var}(\mathbf{Z} \mathbf{v}_i) = \frac{\|\mathbf{Z} \mathbf{v}_i\|^2}{n-1}
= \frac{\langle \mathbf{Z} \mathbf{v}_i, \mathbf{Z} \mathbf{v}_i \rangle}{n-1}
= \left\langle \mathbf{v}_i, \frac{1}{n-1} \mathbf{Z}^T \mathbf{Z} \mathbf{v}_i \right\rangle
= \langle \mathbf{v}_i, \mathbf{R} \mathbf{v}_i \rangle$$

Y dado que $\mathbf{R} \mathbf{v}_i = \lambda_i \mathbf{v}_i$ y $\|\mathbf{v}_i\| = 1$:

$$\langle \mathbf{v}_i, \mathbf{R} \mathbf{v}_i \rangle = \langle \mathbf{v}_i, \lambda_i \mathbf{v}_i \rangle = \lambda_i \langle \mathbf{v}_i, \mathbf{v}_i \rangle = \lambda_i$$

**Por lo tanto, cada eigenvalor $\lambda_i$ representa exactamente la varianza explicada por la $i$-ésima componente principal.**

## Algoritmo paso a paso

1. **Estandarizar** los datos para obtener $\mathbf{Z}$ (media 0, varianza 1).
2. **Calcular la matriz de correlación** $\displaystyle \mathbf{R} = \frac{1}{n-1} \mathbf{Z}^T \mathbf{Z}$.
3. **Calcular eigenvalores** $\lambda_i$ y **eigenvectores** $\mathbf{v}_i$ de $\mathbf{R}$.
4. **Ordenar** los eigenvalores de mayor a menor.
5. **Seleccionar** los $k$ primeros eigenvectores (matriz $\mathbf{Q}_k$ de tamaño $p \times k$).
6. **Proyectar** los datos estandarizados:

$$\mathbf{Y}_{\text{PCA}} = \mathbf{Z} \mathbf{Q}_k$$

La **varianza total** se conserva:

$$\sum_{i=1}^p \lambda_i = \text{traza}(\mathbf{R}) = p$$

pues la traza de la matriz de correlación es igual al número de variables.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Cargar datos de ejemplo (Iris)
iris = load_iris()
X = iris.data
y = iris.target
feature_names = iris.feature_names

# 1. Estandarizar
X_centered = X - X.mean(axis=0)
X_std = X_centered / X.std(axis=0, ddof=1)

# 2. Matriz de covarianza
cov_matrix = np.cov(X_std.T)

# 3. Eigenvalores y eigenvectores
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

# Ordenar de mayor a menor
idx = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

print("Eigenvalores (varianza explicada):", eigenvalues)
print("Porcentaje de varianza explicada:")
print(eigenvalues / eigenvalues.sum() * 100)
```

## Selección del número de componentes

Usamos el **scree plot** y el porcentaje de varianza acumulada.

```python
# Scree plot
plt.figure(figsize=(8,5))
plt.bar(range(1, len(eigenvalues)+1), eigenvalues / eigenvalues.sum() * 100, alpha=0.7, label='Varianza individual')
plt.step(range(1, len(eigenvalues)+1), np.cumsum(eigenvalues / eigenvalues.sum() * 100), color='red', label='Varianza acumulada')
plt.xlabel('Número de componentes')
plt.ylabel('Porcentaje de varianza explicada')
plt.title('Scree Plot - Iris dataset')
plt.legend()
plt.grid(True)
plt.show()

# Varianza acumulada
cum_var = np.cumsum(eigenvalues / eigenvalues.sum())
print("Componentes necesarias para explicar al menos 95% de la varianza:", np.argmax(cum_var >= 0.95) + 1)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_std)

print("Varianza explicada por cada componente:", pca.explained_variance_ratio_)
print("Varianza acumulada:", pca.explained_variance_ratio_.cumsum())

# Visualización 2D
plt.figure(figsize=(8,6))
scatter = plt.scatter(X_pca[:,0], X_pca[:,1], c=y, cmap='viridis')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')
plt.title('Proyección PCA 2D del dataset Iris')
plt.colorbar(scatter, label='Especies')
plt.grid(True)
plt.show()
```

## Interpretación de los componentes

- Los **eigenvectores** (loadings) indican cuánto contribuye cada variable original a cada componente.
- La **Componente 1** suele representar la dirección de máxima varianza ("tamaño general" en muchos conjuntos de datos).
- La **Componente 2** es perpendicular y captura la siguiente fuente de variación.

Ejemplo en Iris: la PC1 suele estar dominada por largo y ancho de pétalos.

## Ventajas

- Reducción de dimensionalidad efectiva.
- Elimina multicolinealidad.
- No requiere etiquetas (no supervisado).
- Fácil de interpretar visualmente.

## Aplicaciones comunes

- Preprocesamiento antes de regresión/clustering/clasificación.
- Compresión de imágenes (Eigenfaces).
- Análisis genómico (miles de genes → pocas componentes).
- Finanzas (análisis de riesgos de portafolios).
- Marketing (segmentación de clientes).
