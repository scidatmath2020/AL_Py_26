**Ejemplo 2.**

Sea $\theta$ un ángulo. Consideremos $T:\mathbb{R}^2\to\mathbb{R}^2$ dada por $$T(a,b)=(a\cos(\theta)-b\sin(\theta),a\sin(\theta)+b\cos(\theta))$$

$$M=\begin{pmatrix}\cos(\theta)&-\sin(\theta)\\\sin(\theta)&\cos(\theta)\end{pmatrix}$$

Sabemos $Ker(T)=\{(0,0)\}$



**Ejemplo 1 de biyectividad**

Sea $T:\mathbb{R}^2\to\mathbb{R}^2$ dada por $$T(x,y)=(x-y,x+y)$$



$T(1,0)=\begin{pmatrix}1\\1\end{pmatrix}$

$T(0,1)=\begin{pmatrix}-1\\1\end{pmatrix}$

$\begin{pmatrix}1&-1\\1&1\end{pmatrix}$

Las dos columnas son linealmente independientes y por lo tanto el rango es 2. Por lo tanto la transformación es **sobreyectiva**.

Por otra, por el teorema de las dimensiones: 

dim(espacio de salida) = dim(kernel) + dim(imagen)

2 = dim(kernel) + 2

dim(kernel)=0

Entonces $T$ es **inyectiva**.


Así, $T$ es **biyectiva**

$T$ es sobreyectiva si y solo si $T$ es inyectiva si y solo $T$ es biyectiva:

Si $T:U\to V$ y $dim(U)=dim(V)=n$ y si además es sobreyectiva, tendremos que $V=Im(T)$ y por lo tanto la dimensión de $Im(T)$ es $n$



Por el Teorema de las dimensiones:
dim(U) = dim(Ker(T)) + dim(Im(T))

$$n=dim(Ker(T))+n$$





Así, $dim(Ker(T))=0$ y entonces $T$ es inyectiva

## Inyectividad es lo mismo que Ker(T)={0}

$\Rightarrow]$ Supongamos que $T$ es inyectiva. Esto significa que si $u\neq v$ entonces $T(u)\neq T(v)$

Si $u\in Ker(T)$ entonces $T(u)=0$; **pero también siempre se cumple que** $T(0)=0$

Si $u\neq 0$ entonces tendríamos $T(u)=0=T(0)$ y esto contradice la inyectividad

$\Leftarrow]$ Supongamos que $Ker(T)=\{0\}$. Veamos que $T$ es inyectiva:

Sean $u$ y $v$ dos vectores con $u\neq v$. Entonces $u-v\neq 0$. LUego, $T(u-v)\neq 0$. Esto implica $T(u)-T(v)\neq 0$


## $T$ es biyectiva si y solo si existe $T^{-1}$ que deshace lo que hace $T$

**Ejemplo 1 de biyectividad**

Sea $T:\mathbb{R}^2\to\mathbb{R}^2$ dada por $$T(x,y)=(x-y,x+y)$$

$T$ es biyectiva.

$T(2,5)=(-3,7)$

Por lo tanto $T^{-1}(-3,7)=(2,5)$

Sea $(a,b)\in\mathbb{R}^2$. Por la biyectividad de $T$, sabemos que existe un único $(x,y)\in\mathbb{R}^2$ tal que $T(x,y)=(a,b)$

Es decir, el sistema de ecuaciones siguiente tiene solución única:

$$\left\{\begin{array}{ccc}x-y&=&a\\x+y&=&b\end{array}\right.$$



$x=\frac{a+b}{2}$ y $y=\frac{b-a}{2}$

Por lo tanto $$T^{-1}(a,b)=\left(\frac{a+b}{2},\frac{b-a}{2}\right)$$

**Teorema**

Si $T:U\to V$ con $dim(U)=dim(V)=n$ es biyectiva, entonces su $[T]_{\beta}^\gamma$ tiene $n$ filas y $n$ columnas. Llamemos $M$ a este matriz.

Entonces $T^{-1}$ y una matriz de representación $M^{-1}$

Y de hecho $MM^{-1}=M^{-1}M=I_n$

## Sistema de ecuaciones

Supongamos un sistema de ecuaciones $Ax=b$, con $A$ cuadrada.

Si $A$ es biyectiva, entonces existe $A^{-1}$. Y por lo tanto

- $Ax=b$ y luego multiplico por $A^{-1}$:

- $A^{-1}Ax=A^{-1}b$

- $Ix=A^{-1}b$

- $x=A^{-1}b$
