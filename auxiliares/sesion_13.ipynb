Tomemos $\beta=\left\{\begin{pmatrix}1\\1\end{pmatrix},\begin{pmatrix}1\\-1\end{pmatrix}\right\}$ base de $\mathbb{R}^2$

Tomemos $\gamma=\left\{\begin{pmatrix}1\\0\\0\end{pmatrix},\begin{pmatrix}1\\1\\0\end{pmatrix},\begin{pmatrix}1\\1\\1\end{pmatrix}\right\}$ base de $\mathbb{R}^3$

Definamos $T:\mathbb{R}^2\to\mathbb{R}^3$ dada por 

$$T(x,y)=(2x,x+y,y)$$



Sea $u\in \mathbb{R}^2$, con $u=(x,y)=\frac{x+y}{2}\begin{pmatrix}1\\1\end{pmatrix}+\frac{x-y}{2}\begin{pmatrix}1\\-1\end{pmatrix}$.

Lo que la teoría me dice es que aplique $T$ a cada elemento de $\beta$:

$$T\begin{pmatrix}1\\1\end{pmatrix}=\begin{pmatrix}2\\2\\1\end{pmatrix},\quad T\begin{pmatrix}1\\-1\end{pmatrix}=\begin{pmatrix}2\\0\\-1\end{pmatrix}$$




Por lo tanto $$T\begin{pmatrix}x\\y\end{pmatrix}=\frac{x+y}{2}\begin{pmatrix}2\\2\\1\end{pmatrix}+\frac{x-y}{2}\begin{pmatrix}2\\0\\-1\end{pmatrix}$$

Por otra parte $$T\begin{pmatrix}1\\1\end{pmatrix}=\begin{pmatrix}2\\2\\1\end{pmatrix}=0\begin{pmatrix}1\\0\\0\end{pmatrix}+1\begin{pmatrix}1\\1\\0\end{pmatrix}+1\begin{pmatrix}1\\1\\1\end{pmatrix}$$

$$T\begin{pmatrix}1\\-1\end{pmatrix}=\begin{pmatrix}2\\0\\-1\end{pmatrix}=2\begin{pmatrix}1\\0\\0\end{pmatrix}+1\begin{pmatrix}1\\1\\0\end{pmatrix}-1\begin{pmatrix}1\\1\\1\end{pmatrix}$$

La matriz que induce es $$[T]_\beta^\gamma=\begin{pmatrix}0&2\\1&1\\1&-1\end{pmatrix}$$

Si yo quiero saber quién es $T(u)$ para cualquier $u$, se tiene que

$$T(u)=[T]_\beta^\gamma[u]_\beta$$





Por ejemplo, en este caso:

$$T\begin{pmatrix}x\\y\end{pmatrix}=\begin{pmatrix}0&2\\1&1\\1&-1\end{pmatrix}\begin{pmatrix}\frac{x+y}{2}\\\frac{x-y}{2}\end{pmatrix}$$
