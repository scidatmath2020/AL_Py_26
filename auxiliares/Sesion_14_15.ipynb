## Kernel

El kernel (o núcleo) de una transformación lineal $T:U\to V$ son aquellos elementos de $V$ tales que $T(v)=0$

**Ejemplo 1.**

$T:\mathrm{R}^3\to\mathbb{R}^2$ dada por $$T(x,y,z)=(x-y,0)$$

Sean $\beta=\{(1,0,0),(0,1,0),(0,0,1)\}$ y $\gamma=\{(1,0),(0,1)\}$

En caso la matriz es directa 
- $T(1,0,0)=(1-0,0)=(1,0)$
- $T(0,1,0)=(0-1,0)=(-1,0)$
- $T(0,0,1)=(0,0)$

Por lo tanto $$[T]_\beta^\gamma=\begin{pmatrix}1&-1&0\\0&0&0\end{pmatrix}$$

Para calcular el kernel buscamos los vectores $u=(x,y,z)$ tales que $[T]_{\beta}^\gamma[u]_{\beta}=\begin{pmatrix}0\\0\end{pmatrix}$

Esto es:
$$\begin{pmatrix}1&-1&0\\0&0&0\end{pmatrix}\begin{pmatrix}x\\y\\z\end{pmatrix}=\begin{pmatrix}0\\0\end{pmatrix}$$


Esto se cumple si y solo si $x-y=0$

Por lo tanto el kernel es $$\mathrm{Ker}(T)=\{(x,x,z)\,:\,x,z\in\mathbb{R}\}$$

Una base es $\beta=\{(1,1,0),(0,0,1)\}$. Además, notemos que la dimensión es 2.

**Ejemplo 2**

Sea $T:\mathbb{R}^4\to\mathbb{R}^2$ dada por 

$$T(x,y,z,t)=(x-z+2t,y+3z-t)$$




El kernel es $\mathrm{Ker}(T)=\{v\in\mathbb{R}^4\,:\,T(v)=0\}$

$T(x,y,zt)=(x-z+2t,y+3z-t)=(0,0)$

$$\left\{\begin{array}{ccc}x-z+2t&=&0\\y+3z-t&=&0\end{array}\right.$$

$x+2t=z$

$y=t-3z=t-3(x+2t)=-3x-5t$

Por lo tanto $$\mathrm{Ker}(T)=\{(x,-3x-5t,x+2t,t)\,:x,t\in\mathbb{R}\}$$

Una base para este kernel es $\beta=\{(0,-5,2,1),(1,-3,1,0)\}$ y su dimensión es 2.

**Ejemplo 3**

Sea $T:\mathbb{R}^3\to\mathbb{R}$ dada por 

$$T(x,y,z)=2x-y+3z$$

$T(x,y,z)=2x-y+3z=0$

Entonces $y=2x+3z$

$$Ker(T)=\{(x,2x+3z,z)\,:\,x,z\in\mathbb{R}\}$$

$\beta=\{(1,2,0),(0,3,1)\}$

**Ejemplo 4**

Sea $T:\mathbb{R}^2\to\mathbb{R}^5$.

$$
T(x, y) = (x + 2y,\; -x,\; 3x - y,\; 4y,\; 2x + y)
$$


$$Ker(T)=\{(0,0)\}$$

**Observación importante:** El espacio vectorial *trivial* es aquel formado únicamente por el vector $\boldsymbol{0}$. Su dimensión es 0

## Imagen
$T:U\to V$

$Im(T)=\{v\in V\,:\,v=T(u)\mbox{ para algún }u\in U\}$


**Ejemplo 1**

$T:\mathrm{R}^3\to\mathbb{R}^2$ dada por $$T(x,y,z)=(x-y,0)$$

El vector $v=(5,2026)$ no está en la imagen de $T$

El vector $v=(a,b)$ está en la imagen de $T$ si y solo existe $(x,y,z)\in\mathbb{R}^3$ tal que $T(x,y,z)=(a,b)$

$(x-y,0)=(a,b)$

Por lo tanto  $x-y=a$ y $0=b$

$$Im(T)=\{(a,0)\,:\,a\in\mathbb{R}\}$$

$T(-a+y,y,z)=(y-(y-a),0=(a,0)$


**Ejemplo 2**

Sea $T:\mathbb{R}^4\to\mathbb{R}^2$ dada por 

$$T(x,y,z,t)=(x-z+2t,y+3z-t)$$


La imagen son los vectores $(a,b)$ tales que $(x-z+2t,y+3z-t)=(a,b)$

$$\left\{\begin{array}{ccc}
x-z+2t&=&a\\
y+3z-t&=&b
\end{array}\right.$$

Anteriormente vimos que $dim(Ker(T))=2$. Pero $T:\mathbb{R}^4\to\mathbb{R}^2$ y $dim(\mathbb{R}^4)=4$. Por el Teorema de las dimensiones $dim(Im(T))=dim(\mathbb{R}^4)-dim(Ker(T))=4-2=2$

Por lo tanto $Im(T)=\mathbb{R}^2$

**Espacio columna de una matriz.**

Sea $M$ una matriz con $n$ columnas; digamos que las columnas son $\mathrm{col}_1$, $\mathrm{col}_2$,...,$\mathrm{col}_n$. El espacio vectorial generado por estas columnas se conoce como *espacio columma de la matriz*.

**Ejemplo.**

$$M=\begin{pmatrix}1&2&3\\0&0&1\\1&2&0\\3&6&4\end{pmatrix}$$

$$COL = \left\{
\mathrm{col}_1=\begin{pmatrix}1\\0\\1\\3\end{pmatrix}, \mathrm{col}_2=\begin{pmatrix}2\\0\\2\\6\end{pmatrix},
\mathrm{col}_3=\begin{pmatrix}3\\1\\0\\4\end{pmatrix}
\right\}$$


El espacio columna es $$span(COL)=\{a_1\mathrm{col}_1+a_2\mathrm{col}_2+a_3\mathrm{col}_3\,:\,a_1,a_2,a_3\mbox{ son números}\}$$

Observemos que hay tres vectores generadores, pero solo dos son linealmente independientes. Por lo tanto, la dimensión del espacio columna en este caso es 2.

**Teorema (importante para calcular la imagen de una transformación lineal)** 

La imagen de una transformación lineal es el espacio columna generado por alguna matriz que la represente.

---

**Ejemplo 1**

$T:\mathrm{R}^3\to\mathbb{R}^2$ dada por $$T(x,y,z)=(x-y,0)$$

Una matriz de representación de $T$ es $$\begin{pmatrix}1&-1&0\\0&0&0\end{pmatrix}$$

Una base para $Im(T)$ es $\gamma = \left\{\begin{pmatrix}1\\0\end{pmatrix}\right\}$ y por lo tanto la dimensión de la imagen es 1. Es decir, el rango es 1.

**Ejemplo 2**

Sea $T:\mathbb{R}^4\to\mathbb{R}^2$ dada por 

$$T(x,y,z,t)=(x-z+2t,y+3z-t)$$

$T(1,0,0,0)=\begin{pmatrix}1\\0\end{pmatrix}$

$T(0,1,0,0)=\begin{pmatrix}0\\1\end{pmatrix}$

$T(0,0,1,0)=\begin{pmatrix}-1\\3\end{pmatrix}$

$T(0,0,0,1)=\begin{pmatrix}2\\-1\end{pmatrix}$


$$\begin{pmatrix}1&0&-1&2\\0&1&3&-1\end{pmatrix}$$

Una base para $Im(T)$ es $$\gamma=\left\{\begin{pmatrix}1\\0\end{pmatrix},\begin{pmatrix}0\\1\end{pmatrix}     \right\}$$

y por lo tanto el rango es 2.

Por ejemplo, si llega el vector $\begin{pmatrix}3\\7\end{pmatrix}$

$$\begin{pmatrix}3\\7\end{pmatrix}=\textcolor{red}{3\begin{pmatrix}1\\0\end{pmatrix}}+\textcolor{blue}{7\begin{pmatrix}0\\1\end{pmatrix}}=\textcolor{red}{3\,T(1,0,0,0)}+\textcolor{blue}{7\,T(0,1,0,0)}=T(3,0,0,0)+T(0,7,0,0)=T(3,7,0,0)$$




**Ejemplo 4**

Sea $T:\mathbb{R}^2\to\mathbb{R}^5$.

$$
T(x, y) = (x + 2y,\; -x,\; 3x - y,\; 4y,\; 2x + y)
$$

$T(1,0)=\begin{pmatrix}1\\-1\\3\\0\\2\end{pmatrix},\quad T(0,1)=\begin{pmatrix}2\\0\\-1\\4\\1\end{pmatrix}$

$$\begin{pmatrix}1&2\\-1&0\\3&-1\\0&4\\2&1\end{pmatrix}$$

$$\gamma = \left\{\begin{pmatrix}0.52\\-0.85\end{pmatrix},\begin{pmatrix}0.85\\0.52\end{pmatrix}\right\}$$
