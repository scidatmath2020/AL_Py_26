Sea $\beta=\{v_1,v_2,...,v_n\}$ una base ortogonal de un espacio vectorial $V$.

Si $v\in V$ es un vector, como $\beta$ es base entonces existen números $a_1,a_2,...,a_n$ tales que $$v=a_1v_1+a_2v_2+...+a_nv_n$$

Fijemos $i\in\{1,2,3,...,n\}$ y observemos el producto $\langle v,v_i\rangle$

\begin{eqnarray*}
&&\langle v,v_i\rangle\\
&=&\langle a_1v_1+a_2v_2+...+a_nv_n,v_i\rangle\\
&=&\langle a_1v_1,v_i\rangle+\langle a_2v_2,v_i\rangle +...+\langle a_nv_n,v_i\rangle\\
&=&a_1\langle v_1,v_i\rangle+a_2\langle v_2,v_i\rangle +...+a_n\langle v_n,v_i\rangle\\
&=&a_i\langle v_i,v_i\rangle\\
&=&a_i\|v_i\|^2
\end{eqnarray*}

Por lo tanto $$a_i=\frac{\langle v,v_i\rangle}{\|v_i\|^2}$$

Esto muestra que $$[v]_\beta=\begin{pmatrix}\frac{\langle v,v_1\rangle}{\|v_1\|^2}\\\frac{\langle v,v_2\rangle}{\|v_2\|^2}\\\vdots\\\frac{\langle v,v_n\rangle}{\|v_n\|^2}\end{pmatrix}$$

**Observación rápida**

Fíjate en el espacio $L^2([-\pi,\pi])=\{f:[-\pi,\pi]\to\mathbb{R}\,|\,\int_{-\pi}^\pi f(x)^2\,\mathrm{d}x<\infty\}$



Con el producto interno $$\langle f,g\rangle =\int_{-\pi}^\pi f(x)g(x)\,\mathrm{d}x$$

Toma la base $\beta=\{1,\sin(t),\cos(t),\sin(2t),\cos(2t),\sin(3t),\cos(3t),...\}$

**Ejemplo**

Expresar al vector $(2,5)$ en la base $\beta=\{(1,2),(-2,1)\}$







$\|(1,2)\|^2=1^2+2^2=5$

$\|(-2,1)\|^2=5$

$\langle (2,5),(1,2)\rangle=2+10=12$

$\langle (2,5),(-2,1)\rangle=-4+5=1$

Entonces $$[(2,5)]_\beta=\begin{pmatrix}\frac{12}{5}\\\frac{1}{5}\end{pmatrix}$$

$\langle (3,5),(1,2)\rangle=3+10=13$

$\langle (3,5),(-2,1)\rangle=-6+5=-1$

Entonces $$[(3,5)]_\beta=\begin{pmatrix}\frac{13}{5}\\\frac{-1}{5}\end{pmatrix}$$

**Observación importante**

Si $\beta=\{v_1,v_2,...,v_n\}$ es base ortogonal, entonces $$\beta^\prime=\left\{\frac{v_1}{\|v_1\|},\frac{v_2}{\|v_2\|},...,\frac{v_n}{\|v_n\|}\right\}$$


es base ortonormal: $$\left\|\frac{\textcolor{red}{v_i}}{\textcolor{blue}{\|v_i\|}}\right\|=\frac{1}{\textcolor{blue}{\|v_i\|}}\|\textcolor{red}{v_i}\|=1$$

## Proceso de Gram Schmidt

Toma un conjunto linealmente independiente $$
\mathbf{v}_1 = \begin{pmatrix} 1 \\ 2 \\ 3 \\ 4 \end{pmatrix}, \qquad
\mathbf{v}_2 = \begin{pmatrix} 2 \\ 1 \\ 0 \\ -1 \end{pmatrix}, \qquad
\mathbf{v}_3 = \begin{pmatrix} 1 \\ 0 \\ 2 \\ 1 \end{pmatrix}
$$




$$u_1 = \frac{v_1}{\|v_1\|}=\frac{1}{\sqrt{30}}\begin{pmatrix} 1 \\ 2 \\ 3 \\ 4 \end{pmatrix}$$

$$u_2=v_2-\frac{\langle v_2,u_1\rangle}{\|u_1\|^2}u_1=\frac{1}{\sqrt{6}}\begin{pmatrix} 2 \\ 1 \\ 0 \\ -1 \end{pmatrix}$$

$$u_3=v_3-\frac{\langle v_3,u_1\rangle}{\|u_1\|}u_1-\frac{\langle v_3,u_2\rangle}{\|u_2\|}u_2=\begin{pmatrix} 1 \\ 0 \\ 2 \\ 1 \end{pmatrix}-\frac{11}{30}\begin{pmatrix} 1 \\ 2 \\ 3 \\ 4 \end{pmatrix}-\frac{1}{6}\begin{pmatrix} 2 \\ 1 \\ 0 \\ -1 \end{pmatrix}=\begin{pmatrix}\frac{3}{10}\\-\frac{9}{10}\\\frac{9}{10}\\-\frac{3}{10}\end{pmatrix}$$
