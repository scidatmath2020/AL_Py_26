- $A$ tiene $m$ filas y $n$ columnas
- $B$ tiene $n$ filas y $p$ columnas

Entonces $AB$ tendrá $m$ filas y $p$ columnas

Algunas propiedades son 
- $(A+B)C=AC+BC$,
- $A(B+C)=AB+AC$
- $AB \neq BA$
- $A(\lambda B)=\lambda AB$

Recordemos el concepto de trasposición de una matriz: $A^T$. Por lo tanto siempre tiene sentido hacer $AA^T$ y $A^TA$.

Además, una propiedad que tiene las traspuestas es que $(AB)^T=B^TA^T$



Con esto en mente $(A\textcolor{red}{A^T})^T=(\textcolor{red}{A^T})^TA^T=AA^T$.

**Recordatorio:** Cuando una matriz $M$ cumple que $M^T=M$, se dice que $M$ es simétrica.

Por lo tanto $AA^T$ y $A^TA$ son matrices simétricas.

Otra propiedad que tienen las matrices es la siguiente: Si $A$ es una matriz de $m$ filas con $n$ columnas y $v$ es un vector de $\mathbb{R}^n$. Pensemos a $v$ como una matriz de una columna con $n$ filas, y entonces tiene sentido hacer
$$Av$$

Además, $Av$ es un matriz de una sola columna pero $m$ filas. Por lo tanto puedo considerar a $Av$ como un vector de $\mathbb{R}^m$


Tomemos $u$ como un vector en $\mathbb{R}^m$. Así que tiene sentido que la operación
$$\langle u,Av\rangle_{\mathbb{R}^m}$$



---

Ahora pensemos en esto: $A^T$ tiene $n$ filas y $m$ columnas. Por lo tanto tiene sentido $A^Tu$ y será una matriz de $n$ filas con una única columna, por lo cual $A^Tu$ se puede pensar como vector de $\mathbb{R}^n$. De esta manera, queda bien definida la operación $$\langle A^Tu,v\rangle_{\mathbb{R}^n}$$




**Teorema del adjunto**

Resulta que si $A$ es una matriz de $m$ filas y $n$ columnas, $u$ es un vector de $\mathbb{R}^m$ y $v$ es un vector de $\mathbb{R}^n$ entonces

$$\langle u,Av\rangle = \langle A^Tu,v\rangle$$
