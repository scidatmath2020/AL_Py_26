## Teorema 1

**Proyección sobre el espacio generado por un vector**

Si $v$ es un vector en $V$, diferente de $0$, y $\langle v\rangle=\mathrm{span}(\{v\})$, entonces la transformación lineal que proyecta todo $V$ sobre $\langle v\rangle$ viene dada por $$P_{\langle v\rangle}u=\frac{\langle u,v\rangle}{\|v\|^2}v$$




## Teorema 2

**Propiedades de columnas con media 0**

Sea $x$ una columna de una tabla, tal que $\overline{x}=0$. Entonces su varianza cumple:

$$Var(x)=\frac{(x_1-\overline{x})^2+(x_2-\overline{x})^2+...+(x_n-\overline{x})^2}{n-1}=\frac{x_1^2+x_2^2...+x_n^2}{n-1}=\frac{\|x\|^2}{n-1}$$

Más aún, si tanto las columnas $x$ como $y$ tienen media 0, entonces
\begin{eqnarray*}
Cov(x,y)&=&\frac{(x_1-\overline{x})(y_1-\overline{y})+(x_2-\overline{x})(y_2-\overline{y})+...+(x_n-\overline{x})(y_n-\overline{y})}{n-1}\\
&=&\frac{x_1y_1+x_2y_2+...+x_ny_n}{n-1}\\
&=&\frac{1}{n-1}\langle x,y\rangle
\end{eqnarray*}

Y por lo tanto 
$$Corr(x,y)=\frac{Cov(x,y)}{\sqrt{Var(x)}\sqrt{Var(y)}}=\frac{\langle x,y\rangle}{\|x\|\|y\|}$$

## Teorema 3

Supongamos que tenemos una matriz $\mathbf{A}$ con $n$ filas y $p$ columnas. Sea $\mathbf{v}$ un vector en $\mathbb{R}^p$ con $\|\mathbf{v}\|=1$.

Supongamos que las filas de $\mathbf{A}$ son $\mathbf{a}^{(1)},\mathbf{a}^{(2)},\dots,\mathbf{a}^{(n)}$. Estas filas son elementos de $\mathbb{R}^p$. Si queremos proyectar el vector $\mathbf{a}^{(k)}$ sobre el espacio generado por $\mathbf{v}$ tenemos:

$$P_{\langle \mathbf{v}\rangle}\mathbf{a}^{(k)} = \frac{\langle \mathbf{a}^{(k)},\mathbf{v}\rangle}{\|\mathbf{v}\|^2} \mathbf{v} = \langle \mathbf{a}^{(k)},\mathbf{v}\rangle \mathbf{v}$$

Por lo tanto, hacer esto con los $n$ renglones de $\mathbf{A}$ equivale a calcular $\mathbf{A}\mathbf{v}$:

$$\mathbf{A}\mathbf{v} = \begin{pmatrix} \langle \mathbf{a}^{(1)},\mathbf{v}\rangle \\ \langle \mathbf{a}^{(2)},\mathbf{v}\rangle \\ \vdots \\ \langle \mathbf{a}^{(n)},\mathbf{v}\rangle \end{pmatrix}$$

## Teorema 4

Sea $\mathbf{B}$ una matriz con $n$ filas y $p$ columnas, donde **cada columna tiene media 0**.

Esto significa que si $B_i$ es la $i$-ésima columna de $\mathbf{B}$, entonces:

$$\frac{B_{1i} + B_{2i} + \dots + B_{ni}}{n} = 0$$

Si tomamos $\mathbf{v} \in \mathbb{R}^p$, entonces $\mathbf{B}\mathbf{v}$ tiene sentido y además:

$$\mathbf{B}\mathbf{v} = \begin{pmatrix} \langle \mathbf{b}^{(1)},\mathbf{v}\rangle \\ \langle \mathbf{b}^{(2)},\mathbf{v}\rangle \\ \vdots \\ \langle \mathbf{b}^{(n)},\mathbf{v}\rangle \end{pmatrix}$$

donde $\mathbf{b}^{(k)}$ es la $k$-ésima **fila** de $\mathbf{B}$. Entonces:

$$\overline{\mathbf{B}\mathbf{v}} = \frac{\langle \mathbf{b}^{(1)},\mathbf{v}\rangle + \dots + \langle \mathbf{b}^{(n)},\mathbf{v}\rangle}{n}
= \left\langle \frac{\mathbf{b}^{(1)}+\dots+\mathbf{b}^{(n)}}{n}, \mathbf{v} \right\rangle = 0$$



### Teorema 5

**Vectores de distancia mínima.**

Sea $V$ un espacio vectorial de **dimensión finita** equipado con un producto interno $ \langle \cdot, \cdot \rangle $. Sea $U \subseteq V $ un **subespacio vectorial**.

Entonces, para todo $ v \in V $, existe un **único** vector $\tilde{v}\in U$  tal que:

$$
\| v - \tilde{v} \|^2 = \min_{u \in U} \| v - u \|^2
$$

y además se cumple que el vector de error es ortogonal a todo el subespacio $U$:

$$
v - \tilde{v} \in U^\perp 
$$

En otras palabras: $$\langle v - \tilde{v}, u \rangle = 0 \mbox{ para todo }u\in U$$

***Demostración***

Dado que $\dim V < \infty$, el subespacio $U$ también tiene dimensión finita. Sea $\{u_1, u_2, \dots, u_k\}$ una **base ortonormal** de $U$.

Definimos:
$$
\tilde{v} = \sum_{i=1}^k \langle v, u_i \rangle u_i
$$

Claramente $\tilde{v} \in U$ porque es combinación lineal de los vectores de la base de $U$.

Sea $w = v - \tilde{v}$. Tomemos cualquier vector $u_j$ de la base:

$$
\langle w, u_j \rangle = \left\langle v - \sum_{i=1}^k \langle v, u_i \rangle u_i, \, u_j \right\rangle = \langle v, u_j \rangle - \sum_{i=1}^k \langle v, u_i \rangle \langle u_i, u_j \rangle
$$

Como la base es ortonormal:

$$
\langle w, u_j \rangle = \langle v, u_j \rangle - \langle v, u_j \rangle = 0
$$

Dado que esto ocurre para todo $u_j$ de la base, y cualquier vector de $U$ es combinación lineal de ellos, se tiene:

$$
\langle w, u \rangle = 0 \quad \forall u \in U
$$

Es decir, $w = v - \tilde{v} \in U^\perp$.

**Ahora veamos que $\tilde{v}$ minimiza la distancia.** Para cualquier $u \in U$, consideremos:

$$
\|v - u\|^2 = \|(v - \tilde{v}) + (\tilde{v} - u)\|^2
$$

Como $v - \tilde{v} \perp (\tilde{v} - u)$ (porque $\tilde{v} - u \in U$), por el teorema de Pitágoras en espacios con producto interno:

$$
\|v - u\|^2 = \|v - \tilde{v}\|^2 + \|\tilde{v} - u\|^2 \geq \|v - \tilde{v}\|^2
$$

La igualdad se da si y solo si $\|\tilde{v} - u\|^2 = 0$, es decir, $u = \tilde{v}$.

Por lo tanto:

$$
\| v - \tilde{v} \|^2 = \min_{u \in U} \| v - u \|^2
$$

## Teorema 6

**Relación entre Imagen y Núcleo de la Transpuesta**

Sea $A: V \to U$ una transformación lineal entre dos espacios vectoriales de dimensión finita equipados con producto interno. Entonces:

$$
\operatorname{Im}(A)^\perp = \operatorname{Ker}(A^T)
$$

donde el complemento ortogonal se toma en $U$ respecto a su producto interno.

***Demostración***

Probaremos las dos inclusiones.

**1. $\operatorname{Im}(A)^\perp \subseteq \operatorname{Ker}(A^T)$**

Sea $w \in \operatorname{Im}(A)^\perp$. Por definición del complemento ortogonal, para cualquier $v\in V$:

$$
0=\langle w, A v \rangle_U =  \langle A^T w, v \rangle_V 
$$


En particular, tomando $v = A^T w$ se obtiene:

$$
\langle A^T w, A^T w \rangle_V = \|A^T w\|_V^2 = 0
$$

lo cual implica $A^T w = 0$, es decir, $w \in \operatorname{Ker}(A^T)$.

**2. $\operatorname{Ker}(A^T) \subseteq \operatorname{Im}(A)^\perp$**

Sea ahora $w \in \operatorname{Ker}(A^T)$, es decir, $A^T w = 0$.

Entonces, para cualquier $v \in V$:

$$
\langle w, A v \rangle_U = \langle A^T w, v \rangle_V = \langle 0, v \rangle_V = 0
$$

Esto significa que $w$ es ortogonal a todo vector de la forma $A v$, es decir, $w$ es ortogonal a todo el subespacio $\operatorname{Im}(A)$. Por definición:

$$
w \in \operatorname{Im}(A)^\perp
$$

Como hemos probado ambas inclusiones, concluimos que:

$$
\operatorname{Im}(A)^\perp = \operatorname{Ker}(A^T)
$$


## Teorema 7 (visto en la sesión 18)

Tomemos una matriz $A$ (no necesariamente cuadrada) donde todas sus columnas son linealmente independientes. Entonces la matriz $A^TA$ es invertible.

***Demostración***

Que una matriz sea invertible equivale a que su kernel es el vector $\{0\}$. Por lo tanto quiero ver que $\mathrm{Ker}(A^TA)=\{0\}$. Es decir, si $u$ es un elemento de $\mathrm{Ker}(A^TA)$, entonces $u=0$.

Sea $u$ un elemento de $\mathrm{Ker}(A^TA)$. Como está en el kernel, $A^TAu=0$

Notemos que $0=\langle u,A^TAu\rangle=\langle Au,Au\rangle=\|Au\|^2$. 

Por lo tanto $Au=0$. Ahora, como las columnas de $A$ son linealmente independientes, tenemos $u=0$



## Teorema 8 

**Relaciones entre eigenvalores de $A^TA$ y $AA^T$**

Supongamos que $A$ tiene $n$ filas y $p$ columnas, donde $p$ es **muy grande** y $n$ es relativamente pequeño.

Entonces $A^TA$ es una matriz cuadrado tiene $p$ filas y $p$ columnas, en tanto que $AA^T$ es una matriz cuadrado tiene $n$ filas y $n$ columnas

Si $\lambda$ es un eigenvalor de $AA^T$ entonces existe un vector diferente de $0$ tal que
$$
\begin{array}{cccc}
&AA^Tv&=&\lambda v\\
\Rightarrow&A^TAA^Tv&=&\lambda A^Tv\\
\Rightarrow&\textcolor{blue}{A^TA}\textcolor{red}{A^Tv}&=&\lambda \textcolor{red}{A^Tv}\\
\end{array}
$$

Esto significa que $A^Tv$ es un eigenvector de $A^TA$, y así $\lambda$ es eigenvalor de $A^TA$. Esto es: **todos los eigenvalores de $AA^T$ son eigenvalores de $A^TA$ y viceversa**
