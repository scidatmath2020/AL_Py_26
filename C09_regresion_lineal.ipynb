# Regresión Lineal y Regresión por Componentes Principales (PCR)

El problema de **regresión lineal** consiste en modelar y predecir una **variable continua** (llamada variable objetivo o variable dependiente) a partir de un conjunto de **variables explicativas** (también llamadas predictores o variables independientes).

### Elementos fundamentales del problema:

- **Variable objetivo ($y$)**: Es la columna que queremos predecir o explicar. Se denomina también "columna objetivo". Es una variable numérica continua (por ejemplo: precio de una casa, ventas mensuales, temperatura, etc.).

- **Predictores ($X$)**: Son las variables que utilizamos para explicar o predecir $y$. Pueden ser una o varias. En notación matricial, $X$ es una matriz de tamaño $n \times p$, donde:
  - $n$ = número de observaciones (filas),
  - $p$ = número de predictores (columnas).

- **Relación entre $y$ y $X$**: Suponemos que existe una relación aproximadamente **lineal** entre los predictores y la variable objetivo. Es decir, que $y$ puede expresarse (aproximadamente) como una combinación lineal de las columnas de $X$ más un término constante.

En forma escalar, el modelo se escribe como:

$$ y_i = \beta_0 + \beta_1 x_{i1} + \beta_2 x_{i2} + \dots + \beta_p x_{ip} + \varepsilon_i \qquad i=1,2,\dots,n $$

donde $\beta_0$ es el **intercepto** y $\beta_1, \beta_2, \dots, \beta_p$ son los coeficientes de regresión.

En notación matricial, el modelo completo se expresa de forma compacta como:

$$ y = \beta_0\begin{pmatrix}1\\1\\\vdots\\1\end{pmatrix}+X\begin{pmatrix}\beta_1 \\\beta_2\\ \vdots \\ \beta_p \end{pmatrix} + \varepsilon $$

Si ahora $[\boldsymbol{1}|X]$ es la matriz $X$ con una primer columna formada únicamente por 1's, y $\beta=\begin{pmatrix}\beta_0\\\beta_1\\\vdots\\\beta_p\end{pmatrix}$ entonces $$y=[\boldsymbol{1}|X]\beta+\varepsilon$$

donde:
- $[\boldsymbol{1}|X]$ es la matriz de predictores **aumentada** con una columna de unos al inicio, de tamaño $n \times (p+1)$,
- $\beta = \begin{pmatrix} \beta_0 \\ \beta_1 \\ \vdots \\ \beta_p \end{pmatrix}$ es el vector de coeficientes de tamaño $(p+1) \times 1$,
- $\varepsilon$ es el vector de errores.

El objetivo final es encontrar una **ecuación** que permita:

1. Entender cómo influye cada predictor en la variable objetivo (interpretación de los coeficientes $\beta_j$).
2. Predecir valores futuros de $y$ cuando se conocen los valores de los predictores.

## Objetivo del método de Mínimos Cuadrados Ordinarios (OLS)

El método más utilizado para estimar los coeficientes $\beta$ es el de **Mínimos Cuadrados Ordinarios (OLS)**.

Su objetivo es encontrar el vector $\hat{\beta}$ que haga que las predicciones $\hat{y} = [\boldsymbol{1}|X]\hat{\beta}$ se acerquen lo más posible a los valores observados $y$. Para lograrlo, se minimiza la **suma de los cuadrados de los residuos** (también llamada suma de cuadrados del error):

$$ 
S(\boldsymbol{b}) = \| y -  [\boldsymbol{1}|X]\boldsymbol{b} \|^2,\quad\mbox{ con }\boldsymbol{b}\in\mathbb{R}^{p+1}
$$

Ahora, notemos que $[\boldsymbol{1}|X]$ es una matriz con $n$ filas y $p+1$ columnas, así que $\mathrm{Im}([\boldsymbol{1}|X])$ es un subespacio de $\mathbb{R}^{n}$. Por el **Teorema 5 de https://github.com/scidatmath2020/AL_Py_26/blob/main/CApendice.ipynb**, para cualquier $v\in\mathbb{R}^{n}$ existe un único $\tilde{v}\in \mathrm{Im}([\boldsymbol{1}|X])$ tal que $$\|v-\tilde{v}\|=\min_{u\in\mathrm{Im}([\boldsymbol{1}|X])}\|v-u\|$$

En particular, para nuestra variable objetivo $y$, que es un elemento de $\mathbb{R}^n$, existe un único $\tilde{y}$ en  $\mathrm{Im}([\boldsymbol{1}|X])$ tal que $$\|y-\tilde{y}\|=\min_{u\in\mathrm{Im}([\boldsymbol{1}|X])}\|y-u\|$$

Es decir, existe un único $\hat{\beta}$ en $\mathbb{R}^{p+1}$ tal que $$\|y-[\boldsymbol{1}|X]\hat{\beta}\|=\min_{u\in\mathrm{Im}([\boldsymbol{1}|X])}\|y-u\|$$

Ese vector $\hat{\beta}$ en $\mathbb{R}^{n+1}$ es el que resuelve nuestro problema. Para calcularlo, usamos de nuevo el **Teorema 5**, que combinado con el **Teorema 6 de https://github.com/scidatmath2020/AL_Py_26/blob/main/CApendice.ipynb** nos dá $$y-[\boldsymbol{1}|X]\hat{\beta}\mbox{ es elemento de }\mathrm{Im}([\boldsymbol{1}|X])^\perp=\mathrm{Ker}([\boldsymbol{1}|X]^T)$$

Por lo tanto $$0=[\boldsymbol{1}|X]^T(y-[\boldsymbol{1}|X]\hat{\beta})=[\boldsymbol{1}|X]^Ty-[\boldsymbol{1}|X]^T[\boldsymbol{1}|X]\hat{\beta}$$

y así $$[\boldsymbol{1}|X]^T[\boldsymbol{1}|X]\hat{\beta}=[\boldsymbol{1}|X]^Ty$$

Estas ecuaciones normales constituyen un sistema de ecuaciones lineales de tamaño $(p+1) \times (p+1)$. En teoría, la solución es
$$ 
\hat{\beta} = ([\boldsymbol{1}|X]^T [\boldsymbol{1}|X])^{-1} [\boldsymbol{1}|X]^T y
$$

Sin embargo, en la práctica esta solución presenta problemas importantes cuando existe **multicolinealidad** (correlaciones altas entre las columnas de $[\boldsymbol{1}|X]$). En esos casos, la matriz $[\boldsymbol{1}|X]^T [\boldsymbol{1}|X]$ puede estar **mal condicionada** o casi singular, lo que provoca:

- Inestabilidad numérica al invertir la matriz,
- Coeficientes $\hat{\beta}$ con muy alta varianza,
- Dificultad para interpretar el efecto individual de cada predictor.

## Uso de PCA en Regresión por Componentes Principales (PCR)

Es importante aclarar un punto clave: El **PCA se aplica únicamente sobre la matriz $X$** (la matriz que contiene solo los predictores), **nunca** sobre la matriz aumentada $[\boldsymbol{1}|X]$.

De esta manera:

1. Se toma **solo la matriz $X$** de predictores (sin la columna de 1's).
2. Se estandarizan las variables para trabajar con la **matriz de correlaciones**:
   $$
   Z_{ij} = \frac{X_{ij} - \bar{X}_j}{s_j}
   $$
   donde $\bar{X}_j$ y $s_j$ son la media y desviación estándar de la columna $j$.
3. Se aplica PCA sobre la matriz estandarizada $Z$ y se obtienen las **componentes principales** $Y$:
   $$
   Y = Z Q
   $$
   donde $Q$ es la matriz de autovectores de la matriz de correlaciones de $Z$.

Las columnas de $Y$ son **ortogonales** entre sí.



### Modelo de regresión en el espacio de componentes principales

Una vez obtenida $Y$, el modelo de regresión lineal se reformula como:

$$
y = [\boldsymbol{1}|Y] \gamma + \varepsilon
$$

donde $\gamma = (\gamma_0,\gamma_1, \gamma_2, \dots, \gamma_p)^T$ es el vector de coeficientes en el nuevo espacio.

### Resolución de las ecuaciones normales en el espacio de componentes principales

Aplicando el método de mínimos cuadrados al modelo anterior, las ecuaciones normales quedan:

$$
[\boldsymbol{1}|Y]^T [\boldsymbol{1}|Y] \hat{\gamma} = [\boldsymbol{1}|Y]^T y
$$

Dado que el PCA se realizó sobre la matriz de correlaciones, las columnas de $Y$ tienen media cero. Por lo tanto, la primera columna de $[\boldsymbol{1}|Y]$ (el vector de unos) es **ortogonal** a todas las columnas de $Y$:

$$\langle\boldsymbol{1},Y_i\rangle=1\cdot Y_{1i}+1\cdot Y_{2i}+...+1\cdot Y_{ni}=Y_{1i}+\cdot Y_{2i}+...+\cdot Y_{ni}=0$$

Así, las columnas de $[\boldsymbol{1}|Y]$ son linealmente independientes, y por el **Teorema 7 de https://github.com/scidatmath2020/AL_Py_26/blob/main/CApendice.ipynb** tenemos que $$[\boldsymbol{1}|Y]^T[\boldsymbol{1}|Y]$$ es invertible.

Luego, $$\hat{\gamma}=([\boldsymbol{1}|Y]^T [\boldsymbol{1}|Y])^{-1} [\boldsymbol{1}|Y]^T y$$

### Recuperación de los coeficientes originales en el espacio de $X$

Una vez obtenidos los coeficientes $\hat{\gamma}$ en el espacio de las componentes principales, es posible recuperar los coeficientes originales correspondientes a las variables de $X$.

Recordemos que $Y = Z Q$, donde $Z$ es la matriz de predictores estandarizados y $Q$ es la matriz de autovectores.

Dado que el modelo en el espacio de componentes es

$$
y = [\boldsymbol{1}|Y] \gamma + \varepsilon
$$

y $Y = Z Q$, podemos escribir:

$$
y = \gamma_0 \boldsymbol{1} + Z Q \gamma_{1:p} + \varepsilon
$$

donde $\gamma_{1:p} = (\gamma_1, \gamma_2, \dots, \gamma_p)^T$ es la parte de $\gamma$ sin el intercepto.

Como $Z = (X - \bar{X}) D^{-1}$, donde $D$ es la matriz diagonal que contiene las desviaciones estándar $s_j$ de cada variable (es decir, $D = \text{diag}(s_1, s_2, \dots, s_p)$), entonces:

$$
y = \gamma_0 \boldsymbol{1} + (X - \bar{X}) D^{-1} Q \gamma_{1:p} + \varepsilon
$$

Reordenando, obtenemos la expresión en términos de los predictores originales $X$:

$$
y = \left[ \gamma_0 - (\bar{X} D^{-1} Q \gamma_{1:p})^T \boldsymbol{1} \right] + X \left( D^{-1} Q \gamma_{1:p} \right) + \varepsilon
$$

Por lo tanto, los coeficientes originales $\hat{\beta}$ asociados a las variables de $X$ se recuperan como:

$$
\hat{\beta} = D^{-1} Q \hat{\gamma}_{1:p}
$$

donde:
- $\hat{\beta} = (\hat{\beta}_1, \hat{\beta}_2, \dots, \hat{\beta}_p)^T$,
- $\hat{\gamma}_{1:p} = (\hat{\gamma}_1, \hat{\gamma}_2, \dots, \hat{\gamma}_p)^T$ son los coeficientes correspondientes a las componentes principales (sin el intercepto),
- $D^{-1} = \text{diag}(1/s_1, 1/s_2, \dots, 1/s_p)$.

El intercepto original $\hat{\beta}_0$ se calcula como:

$$
\hat{\beta}_0 = \bar{y} - \bar{X}^T \hat{\beta}
$$
