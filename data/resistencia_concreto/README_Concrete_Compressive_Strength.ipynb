# Concrete Compressive Strength Dataset

## Descripción general
Este dataset proviene del **UCI Machine Learning Repository** y contiene información de **930 muestras de concreto** probadas en laboratorio.  

El objetivo es predecir la **resistencia a la compresión** del concreto (una propiedad clave en ingeniería civil) en función de sus ingredientes y el tiempo de curado.

Es un dataset excelente para **regresión lineal múltiple**, **PCA** (debido a las correlaciones entre componentes) y modelos no lineales, ya que la relación entre los ingredientes y la resistencia es altamente no lineal.

- **Número de filas**: 1,030
- **Número de columnas**: 9 (8 predictoras + 1 objetivo)
- **Valores faltantes**: Ninguno
- **Tipo**: Todas las variables son numéricas

## Columnas del dataset

| Columna                  | Nombre completo                  | Descripción                                                                 | Tipo de dato | Unidad / Escala                          |
|--------------------------|----------------------------------|-----------------------------------------------------------------------------|--------------|------------------------------------------|
| **Cement**               | Cement (component 1)             | Cantidad de cemento en la mezcla                                            | numérica     | kg/m³                                    |
| **BlastFurnaceSlag**     | Blast Furnace Slag (component 2) | Cantidad de escoria de alto horno                                           | numérica     | kg/m³                                    |
| **FlyAsh**               | Fly Ash (component 3)            | Cantidad de ceniza volante                                                  | numérica     | kg/m³                                    |
| **Water**                | Water (component 4)              | Cantidad de agua                                                            | numérica     | kg/m³                                    |
| **Superplasticizer**     | Superplasticizer (component 5)   | Cantidad de superplastificante (aditivo reductor de agua)                   | numérica     | kg/m³                                    |
| **CoarseAggregate**      | Coarse Aggregate (component 6)   | Cantidad de agregado grueso (grava)                                         | numérica     | kg/m³                                    |
| **FineAggregate**        | Fine Aggregate (component 7)     | Cantidad de agregado fino (arena)                                           | numérica     | kg/m³                                    |
| **Age**                  | Age                              | Tiempo de curado del concreto después de la mezcla                          | numérica     | Días (generalmente de 1 a 365)           |
| **Strength**             | Concrete Compressive Strength    | Resistencia a la compresión del concreto (variable objetivo)                | numérica     | MPa (Megapascales)                       |

## Notas importantes

- Todas las variables de ingredientes (**Cement** hasta **FineAggregate**) están expresadas en **kilogramos por metro cúbico** (kg/m³) de mezcla.
- **Age** es la única variable que no está en kg/m³; representa los días de curado.
- La **Strength** (resistencia) se mide en **Megapascales (MPa)**. Valores típicos van desde ~10 MPa hasta ~80-90 MPa.
- Este dataset es ideal para aplicar **PCA sobre la matriz de correlaciones** porque varios componentes (cemento, agua, agregados, etc.) están correlacionados entre sí.
- La relación entre las variables predictoras y la resistencia **no es lineal**, por lo que después de PCA + regresión lineal, suele ser útil probar modelos más avanzados (Random Forest, XGBoost, etc.).
- No contiene variables categóricas → perfecto para estandarizar directamente y aplicar PCA.
