# Advertising Dataset

## Descripción general
Este dataset contiene información sobre el **presupuesto invertido en publicidad** en tres medios diferentes y las **ventas** generadas por una empresa en 175 zonas geográficas diferentes.

Es uno de los datasets más clásicos y utilizados para introducir **regresión lineal múltiple**. Con solo 175 observaciones y 4 columnas, es ideal para entender la relación entre inversión publicitaria y resultados de ventas.

- **Número de filas**: 175
- **Número de columnas**: 4 (3 predictoras + 1 objetivo)
- **Valores faltantes**: Ninguno
- **Tipo**: Todas las variables son numéricas

## Columnas del dataset

| Columna     | Nombre completo         | Descripción                                                              | Tipo de dato | Unidad / Escala             |
|-------------|-------------|-------------|-------------|-------------|
| TV          | TV Advertising          | Presupuesto invertido en publicidad en televisión                        | numérica     | Miles de dólares        |
| Radio       | Radio Advertising       | Presupuesto invertido en publicidad en radio                             | numérica     | Miles de dólares        |
| Newspaper   | Newspaper Advertising   | Presupuesto invertido en publicidad en periódicos                        | numérica     | Miles de dólares       |
| Sales       | Sales                   | Ventas generadas (variable objetivo)                                     | numérica     | Miles de unidades           |



## Notas importantes

- Todas las variables están expresadas en **miles de dólares** para las inversiones publicitarias (`TV`, `Radio`, `Newspaper`).
- La variable objetivo **`Sales`** representa las ventas en **miles de unidades** del producto.
- Este dataset es muy limpio y no contiene valores atípicos extremos ni valores faltantes.
- Existe una **fuerte relación lineal** entre la inversión en TV y las ventas, mientras que Radio y Newspaper tienen una relación más débil.
- Es excelente para practicar:
  - Regresión lineal múltiple
  - Análisis de multicolinealidad (aunque es baja en este caso)
  - Visualización de relaciones (scatter plots)
  - Comparación entre regresión simple y múltiple
- **PCA**: Aunque se puede aplicar, con solo 3 variables predictoras el beneficio de reducción de dimensionalidad es limitado. Sin embargo, es útil para demostrar el concepto.
