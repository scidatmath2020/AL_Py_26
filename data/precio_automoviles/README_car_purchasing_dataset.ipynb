# Car Purchasing Data Dataset

## Descripción general
Este dataset contiene información de **450 clientes** y el monto que cada uno gastó (o está dispuesto a gastar) en la compra de un automóvil nuevo.

El objetivo principal es predecir el **Car Purchase Amount** (monto de compra del auto) utilizando características demográficas y financieras de los clientes.  

Es un dataset muy bueno para practicar **regresión lineal múltiple**, manejo de variables categóricas y **PCA** (después de codificar las variables categóricas).

- **Número de filas**: 450
- **Número de columnas**: 9 (8 predictoras + 1 objetivo)
- **Valores faltantes**: Ninguno
- **Tipo**: Mezcla de variables numéricas y categóricas

## Columnas del dataset

| Columna              | Nombre completo            | Descripción                                                                 | Tipo de dato | Unidad / Escala                          |
|----------------------|----------------------------|-----------------------------------------------------------------------------|--------------|------------------------------------------|
| **Customer Name**    | Customer Name              | Nombre del cliente                                                          | texto        | -                                        |
| **Customer e-mail**  | Customer Email             | Correo electrónico del cliente                                              | texto        | -                                        |
| **Country**          | Country                    | País de residencia del cliente                                              | categórica   | Nombre del país                          |
| **Gender**           | Gender                     | Género del cliente                                                          | categórica   | Male / Female                            |
| **Age**              | Age                        | Edad del cliente                                                            | numérica     | Años                                     |
| **Annual Salary**    | Annual Salary              | Salario anual del cliente                                                   | numérica     | Dólares                               |
| **Credit Card Debt** | Credit Card Debt           | Monto de deuda en tarjeta de crédito                                        | numérica     | Dólares                               |
| **Net Worth**        | Net Worth                  | Patrimonio neto del cliente (activos menos pasivos)                         | numérica     | Dólares                               |
| **Car Purchase Amount** | Car Purchase Amount     | Monto gastado en la compra del automóvil (variable objetivo)                | numérica     | Dólares                               |

## Notas importantes

- Las columnas **Customer Name** y **Customer e-mail** son identificadores y normalmente se eliminan antes de modelar (no aportan información predictiva).
- **Country** y **Gender** son variables categóricas. Para incluirlas en regresión o PCA es necesario aplicar **One-Hot Encoding** (variables dummy).
- Las variables numéricas (**Age**, **Annual Salary**, **Credit Card Debt**, **Net Worth**) suelen tener diferentes escalas, por lo que es **obligatorio estandarizar** antes de aplicar PCA.
- La variable objetivo **`Car Purchase Amount`** es continua y representa el monto total que el cliente pagó (o está dispuesto a pagar) por el auto.
- Este dataset es ideal para demostrar **multicolinealidad** entre variables financieras (Annual Salary y Net Worth suelen estar correlacionadas).
