# Fish Market Dataset

## Descripción general
Este dataset contiene información física de **144 peces** pertenecientes a **7 especies diferentes** comercializadas en un mercado de pescado.  

El objetivo principal es predecir el **peso** del pescado (`Weight`) a partir de sus medidas corporales (longitudes, altura y ancho).  

Es un dataset muy bueno para practicar **regresión lineal múltiple**, feature engineering (las tres longitudes están altamente correlacionadas) y **PCA**, ya que las variables de longitud suelen presentar multicolinealidad fuerte.

- **Número de filas**: 144
- **Número de columnas**: 7 (6 predictoras + 1 objetivo)
- **Valores faltantes**: Ninguno
- **Tipo**: 1 variable categórica + 6 variables numéricas

## Columnas del dataset

| Columna       | Nombre completo              | Descripción                                                                 | Tipo de dato | Unidad / Escala                  |
|---------------|------------------------------|-----------------------------------------------------------------------------|--------------|----------------------------------|
| **Species**   | Species                      | Especie del pez (variable categórica)                                       | categórica   | 7 categorías (Bream, Roach, Whitefish, Parkki, Perch, Pike, Smelt) |
| **Weight**    | Weight                       | Peso del pez (variable objetivo)                                            | numérica     | Gramos (g)                       |
| **Length1**   | Vertical Length              | Longitud vertical del pez                                                   | numérica     | Centímetros (cm)                 |
| **Length2**   | Diagonal Length              | Longitud diagonal del pez                                                   | numérica     | Centímetros (cm)                 |
| **Length3**   | Cross Length                 | Longitud transversal (de la cabeza a la cola)                               | numérica     | Centímetros (cm)                 |
| **Height**    | Height                       | Altura del pez (medida perpendicular al cuerpo)                             | numérica     | Centímetros (cm)                 |
| **Width**     | Width                        | Ancho del pez (grosor máximo)                                               | numérica     | Centímetros (cm)                 |

## Notas importantes

- Las tres variables de longitud (**Length1**, **Length2** y **Length3**) están muy correlacionadas entre sí. Esto hace que el dataset sea ideal para aplicar **PCA sobre la matriz de correlaciones** y reducir la multicolinealidad.
- **Species** es una variable categórica con 7 clases. Puedes usarla creando variables dummy (one-hot encoding) si quieres incluirla en el modelo, o trabajar solo con las variables numéricas.
- El **Weight** (peso) es la variable objetivo y es claramente continua.
- Algunas observaciones de la especie **Smelt** tienen pesos muy bajos, y la especie **Pike** tiende a ser más larga y delgada.
- Este dataset es excelente para demostrar cómo PCA ayuda a combinar las medidas de longitud en un solo componente "tamaño general".
