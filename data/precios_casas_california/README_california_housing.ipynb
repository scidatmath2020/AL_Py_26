# California Housing Dataset

## Descripción general
Este dataset proviene del censo de Estados Unidos de 1990 y contiene información sobre **20,140 bloques censales** (block groups) en California.  

El objetivo principal es predecir el **precio mediano de las viviendas** (`PRICE`) en cada bloque utilizando características demográficas, de vivienda y geográficas.

Todas las variables (excepto la variable objetivo) se utilizan como predictoras en modelos de regresión.

## Columnas del dataset

| Columna      | Nombre completo         | Descripción breve                                              | Unidad / Escala                          |
|--------------|-------------------------|----------------------------------------------------------------|------------------------------------------|
| **MedInc**   | Median Income           | Ingreso mediano de los hogares                                 | Decenas de miles de dólares              |
| **HouseAge** | Median House Age        | Edad mediana de las viviendas                                  | Años                                     |
| **AveRooms** | Average Rooms           | Promedio de habitaciones por hogar                             | Habitaciones/hogar                       |
| **AveBedrms**| Average Bedrooms        | Promedio de dormitorios por hogar                              | Dormitorios/hogar                        |
| **Population**| Population             | Población total del bloque                                     | Personas                                 |
| **AveOccup** | Average Occupancy       | Promedio de personas por hogar                                 | Personas/hogar                           |
| **Latitude** | Latitude                | Latitud del bloque censal                                      | Grados                                   |
| **Longitude**| Longitude               | Longitud del bloque censal                                     | Grados (Oeste)                           |
| **PRICE**    | Median House Value      | Precio mediano de la vivienda (variable objetivo)              | Cientos de miles de dólares              |

## Notas importantes

- **AveRooms**, **AveBedrms** y **AveOccup** son **promedios por hogar**, no por vivienda individual. En zonas con muchas segundas residencias, hoteles o casas vacías, estos valores pueden ser altos o atípicos.
- **Latitude** y **Longitude** capturan el efecto de ubicación geográfica. En California, la proximidad a la costa y a grandes ciudades (Bay Area, Los Ángeles) tiene gran influencia en los precios.
- El dataset **no contiene valores faltantes**.
- **Unidades del target (PRICE)**: El valor está expresado en **cientos de miles de dólares**. El valor máximo suele estar truncado en $500,000 (5.0).

## Uso recomendado
- Este dataset es ideal para practicar **regresión lineal**, regresión múltiple, PCA, feature engineering y análisis espacial.
- Se recomienda **estandarizar** todas las variables antes de aplicar PCA (especialmente porque estás trabajando sobre la matriz de correlaciones).
