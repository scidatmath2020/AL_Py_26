# Matrices especiales

En este capítulo estudiaremos matrices con propiedades estructurales que las hacen especialmente útiles en aplicaciones como estadística, machine learning y geometría computacional.

## Matrices simétricas

**Definición:** Una matriz cuadrada $A \in \mathbb{R}^{n \times n}$ es **simétrica** si es igual a su transpuesta:
$$A = A^\top$$

Equivalentemente, $a_{ij} = a_{ji}$ para todos $i, j$.

**Ejemplo:**
$$A = \begin{pmatrix} 4 & 1 & -2 \\ 1 & 3 & 5 \\ -2 & 5 & 6 \end{pmatrix}$$

Verificar que $A = A^\top$.

**Propiedades fundamentales:**

1. **Eigenvalues reales:** Todos los valores propios de una matriz simétrica son reales.
2. **Eigenvectors ortogonales:** Los vectores propios correspondientes a eigenvalues distintos son ortogonales.
3. **Diagonalización ortogonal:** Toda matriz simétrica puede escribirse como:
   $$A = Q \Lambda Q^\top$$
   donde $Q$ es ortogonal (sus columnas son eigenvectors ortonormales) y $\Lambda$ es diagonal con los eigenvalues.
4. **Formas cuadráticas:** Para cualquier vector $x$, la forma cuadrática $x^\top A x$ es un escalar. Si $A$ es simétrica, esta forma cuadrática tiene propiedades especiales.

## Matrices semidefinidas positivas

**Definición:** Una matriz simétrica $A \in \mathbb{R}^{n \times n}$ es **semidefinida positiva** (SDP) si para todo vector $x \in \mathbb{R}^n$:
$$\langle x,A x\rangle \geq 0$$

Es **definida positiva** si $\langle x,A x\rangle > 0$ para todo $x \neq 0$.

**Caracterizaciones equivalentes:**

1. **Eigenvalues:** Todos los eigenvalues de $A$ son $\geq 0$ (positivos si es definida positiva)
2. **Cholesky:** Existe una matriz $L$ (triangular inferior) tal que $A = L L^\top$
3. **Raíz cuadrada:** Existe una matriz simétrica $B$ tal que $A = B^2$
4. **Forma cuadrática:** Para cualquier matriz $X$, $X^\top X$ es semidefinida positiva

**Propiedades:**

1. **Suma:** La suma de matrices SDP es SDP
2. **Producto por escalar:** Si $A$ es SDP y $\alpha \geq 0$, entonces $\alpha A$ es SDP
3. **Traza:** $\text{tr}(A) \geq 0$ y es igual a la suma de eigenvalues
4. **Determinante:** $\det(A) \geq 0$ (producto de eigenvalues)
5. **Menores principales:** Todos los menores principales son no negativos (criterio de Sylvester)

**Ejemplos:**

$$A = \begin{pmatrix} 2 & 1 \\ 1 & 2 \end{pmatrix}$$

Verifiquemos que es definida positiva:
- Eigenvalues: $\lambda_1 = 3 > 0$, $\lambda_2 = 1 > 0$
- Forma cuadrática: $x^\top A x = 2x_1^2 + 2x_2^2 + 2x_1x_2 = 2(x_1^2 + x_1x_2 + x_2^2) \geq 0$

$$B = \begin{pmatrix} 1 & 1 \\ 1 & 1 \end{pmatrix}$$

Esta es semidefinida positiva (no definida positiva):
- Eigenvalues: $\lambda_1 = 2 > 0$, $\lambda_2 = 0$
- Para $x = (1, -1)$: $x^\top B x = 0$

**Importancia en aplicaciones:**
- **Matriz de covarianza:** Es siempre semidefinida positiva
- **Hessiana:** En optimización, si es definida positiva indica un mínimo local
- **Métricas:** Definen productos internos y distancias
- **Kernel methods:** Los kernels son funciones que generan matrices SDP

## Matrices de proyección ortogonal

**Definición:** Una matriz $P \in \mathbb{R}^{n \times n}$ es una **proyección ortogonal** si cumple:
1. **Idempotencia:** $P^2 = P$ (proyectar dos veces es lo mismo que proyectar una vez)
2. **Simetría:** $P^\top = P$ (la proyección es ortogonal)

**Interpretación geométrica:** Una matriz de proyección ortogonal proyecta vectores sobre un subespacio $S$ a lo largo del complemento ortogonal $S^\perp$.

**Propiedad:** Si $P$ es una proyección ortogonal sobre un subespacio $S$, entonces $I - P$ es la proyección ortogonal sobre $S^\perp$.

**Caso particular 1: Proyección sobre el espacio columna de una matriz**

Si $X \in \mathbb{R}^{n \times p}$ tiene columnas linealmente independientes, la matriz que proyecta ortogonalmente sobre el espacio columna de $X$ es:
$$P = X (X^\top X)^{-1} X^\top$$

**Caso particular 2: Proyección sobre el espacio generado por un vector**

Si $v$ es un vector en $V$, diferente de $0$, y $\langle v\rangle=\mathrm{span}(\{v\})$, entonces la transformación lineal que proyecta todo $V$ sobre $\langle v\rangle$ viene dada por $$P_{\langle v\rangle}u=\frac{\langle u,v\rangle}{\|v\|^2}v$$

**Ejemplo: Proyección en $\mathbb{R}^2$ sobre una recta**

## Matrices ortogonales

**Definición:** Una matriz cuadrada $Q \in \mathbb{R}^{n \times n}$ es **ortogonal** si:
$$Q^\top Q = Q Q^\top = I$$

Es decir, su inversa es su transpuesta: $Q^{-1} = Q^\top$.

**Propiedades:**

1. **Preservan norma:** $\|Qx\| = \|x\|$ para todo $x$ (isometrías)
2. **Preservan producto punto:** $\langle Qx,Qy\rangle = \langle x , y\rangle$
3. **Columnas ortonormales:** Las columnas de $Q$ forman una base ortonormal de $\mathbb{R}^n$
4. **Filas ortonormales:** Las filas también forman una base ortonormal
5. **Determinante:** $\det(Q) = \pm 1$
   - Si $\det(Q) = 1$, es una **rotación**
   - Si $\det(Q) = -1$, es una **reflexión**

**Ejemplo 1: Matriz de rotación en $\mathbb{R}^2$**
$$R(\theta) = \begin{pmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{pmatrix}$$

**Ejemplo 2: Matriz de reflexión en $\mathbb{R}^2$ sobre el eje x**
$$H = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}$$

## Matriz de covarianza

En estadística y machine learning, la matriz de covarianza mide cómo se relacionan linealmente distintas variables.

**Datos centrados:** Dada una matriz de datos $X \in \mathbb{R}^{n \times p}$ donde $n$ es el número de observaciones y $p$ el número de variables, primero **centramos** los datos restando la media de cada columna:
$$\tilde{X} = X - \mathbf{1}\mu^\top$$
donde $\mu \in \mathbb{R}^p$ es el vector de medias de cada variable.

**Importante.** Sea $x$ una columna de una tabla, tal que $\overline{x}=0$. Entonces su varianza cumple:

$$Var(x)=\frac{(x_1-\overline{x})^2+(x_2-\overline{x})^2+...+(x_n-\overline{x})^2}{n-1}=\frac{x_1^2+x_2^2...+x_n^2}{n-1}=\frac{\|x\|^2}{n-1}$$


**Definición:** La **matriz de covarianza muestral** es:
$$S = \frac{1}{n-1} \tilde{X}^\top \tilde{X}$$

**Propiedades:**

1. **Simétrica:** $S^\top = S$
2. **Semidefinida positiva:** $\langle x, S x\rangle \geq 0$ para todo $x$
3. Los elementos diagonales $S_{jj}$ son las varianzas de cada variable
4. Los elementos fuera de la diagonal $S_{ij}$ son las covarianzas entre variables

**Interpretación:** $S_{ij} = \text{Cov}(x_i, x_j) = \frac{1}{n-1} \sum_{k=1}^n (x_{ki} - \mu_i)(x_{kj} - \mu_j)$
