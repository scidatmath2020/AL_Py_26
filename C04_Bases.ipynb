# Base de un espacio vectorial

**Definición:** Un conjunto $\beta = \{v_1, v_2, \dots, v_n\}$ de vectores en un espacio vectorial $V$ es una **base** si cumple dos condiciones:

1. **Independencia lineal:** La única combinación lineal que da el vector cero es la trivial:
   $$c_1v_1 + c_2v_2 + \dots + c_nv_n = 0 \quad \Rightarrow \quad c_1 = c_2 = \dots = c_n = 0$$

2. **Generador:** Todo vector $v \in V$ puede expresarse como combinación lineal de los vectores de $\beta$:
   $$v = a_1v_1 + a_2v_2 + \dots + a_nv_n$$

Los escalares $a_1, a_2, \dots, a_n$ son únicos y se llaman **coordenadas** de $v$ en la base $\beta$. Los denotamos como:
$$[v]_{\beta} = \begin{pmatrix} a_1 \\ a_2 \\ \vdots \\ a_n \end{pmatrix}$$

**Dimensión:** El número $n$ de vectores en una base es la **dimensión** del espacio, denotada $\dim(V)$.

## Ejemplos de bases

**Ejemplo 1. Base canónica de $\mathbb{R}^n$**

La base más sencilla: $e_1 = (1,0,\dots,0)$, $e_2 = (0,1,\dots,0)$, ..., $e_n = (0,0,\dots,1)$.

En esta base, las coordenadas de un vector son sus propias componentes:
$$[(x_1,x_2,\dots,x_n)]_{\text{can}} = \begin{pmatrix} x_1 \\ x_2 \\ \vdots \\ x_n \end{pmatrix}$$

**Ejemplo 2.**

Ahora tomemos $\beta = \{(1,1), (1,-1)\}$. En esta base, el vector $(2,3)$ tiene coordenadas que debemos calcular.

**Solución.**

Como $\mathbb{R}^2$ tiene dimensión 2, queremos hallar escalares $a_1$ y $a_2$ tales que $$\begin{pmatrix}2\\3\end{pmatrix}=a_1\begin{pmatrix}1\\1\end{pmatrix}+a_2\begin{pmatrix}1\\-1\end{pmatrix}$$

Es decir, resolver el sistema de ecuaciones $$\left\{\begin{array}{ccc}a_1+a_2&=&2\\a_1-a_2&=&3\end{array}\right.$$

Al resolver el sistema obtenemos $a_1=2.5$ y $a_2=-0.5$. Por lo tanto $$[(2,3)]_\beta=\begin{pmatrix}2.5\\-0.5\end{pmatrix}$$

**Ejemplo 3.**

En el espacio vectorial $\mathbb{R}^3$ tomemos $$\beta=\{(1, 2, 1), (0, -1, 3), (2, 1, -1)\}$$

Sea el vector $u = (5, -1, 4)$. Queremos expresar $u$ en términos de $\beta$. Es decir, hallar escalares $a_1$, $a_2$, $a_3$ tales que $$\begin{pmatrix}5\\-1\\4\end{pmatrix}=a_1\begin{pmatrix}1\\2\\1\end{pmatrix}+a_2\begin{pmatrix}0\\-1\\3\end{pmatrix}+a_3\begin{pmatrix}2\\1\\-1\end{pmatrix}$$

Es decir, resolver el sistema de ecuaciones $$\left\{\begin{array}{ccc}a_1\quad+2a_3&=&5\\2a_1-a_2+a_3&=&-1\\a_1+3a_2-a_3&=&4\end{array}\right.$$

Al resolver el sistema obtenemos $a_1=-0.667$, $a_2=2.5$ y $a_3=2.833$. Por lo tanto $$[(5,-1,4)]_\beta=\begin{pmatrix}-0.667\\2.5\\2.833\end{pmatrix}$$

```python
# Cálculo de coordenadas en Python

import numpy as np

base = [[1,2,1],[0,-1,3],[2,1,-1]]
v = np.array([[5],
              [-1],
              [4]
             ])
A = np.column_stack(base)

np.linalg.solve(A, v)
```

## Bases ortogonales y ortonormales

**Definición (Base ortogonal):** Una base $\beta = \{v_1, v_2, \dots, v_n\}$ de un espacio vectorial $V$ con producto interno es **ortogonal** si sus vectores son ortogonales entre sí:
$$\langle v_i , v_j\rangle = 0 \quad \text{para todo } i \neq j$$

**Definición (Base ortonormal):** Una base $\beta$ es **ortonormal** si es ortogonal y además cada vector tiene norma igual a 1:
$$\langle v_i , v_j \rangle = \begin{cases} 1 & \text{si } i = j \\ 0 & \text{si } i \neq j \end{cases}$$

**Propiedad fundamental:** En una base ortogonal, las coordenadas de un vector se calculan mediante proyecciones. Si $\beta = \{v_1, \dots, v_n\}$ es una base ortogonal, entonces:
$$[v]_{\beta} = \begin{pmatrix} \frac{v \cdot v_1}{\|v_1\|^2} \\ \frac{v \cdot v_2}{\|v_2\|^2} \\ \vdots \\ \frac{v \cdot v_n}{\|v_n\|^2} \end{pmatrix}$$

Si la base es ortonormal, la expresión se simplifica:
$$[v]_{\beta} = \begin{pmatrix} v \cdot v_1 \\ v \cdot v_2 \\ \vdots \\ v \cdot v_n \end{pmatrix}$$

**Ejemplo 4. Base ortogonal en $\mathbb{R}^2$**

Sea $\beta = \{(1,2), (-2,1)\}$. Verifiquemos que es ortogonal:
$$\langle (1,2) , (-2,1)\rangle = 1(-2) + 2(1) = -2 + 2 = 0$$

Ahora encontremos las coordenadas del vector $v = (3,5)$ en esta base:
$$a_1 = \frac{v \cdot v_1}{\|v_1\|^2} = \frac{(3,5) \cdot (1,2)}{1^2+2^2} = \frac{3+10}{5} = \frac{13}{5} = 2.6$$
$$a_2 = \frac{v \cdot v_2}{\|v_2\|^2} = \frac{(3,5) \cdot (-2,1)}{(-2)^2+1^2} = \frac{-6+5}{5} = \frac{-1}{5} = -0.2$$

Por lo tanto $[(3,5)]_\beta = \begin{pmatrix} 2.6 \\ -0.2 \end{pmatrix}$

**Ejemplo 5. Base ortonormal en $\mathbb{R}^3$**

Sea $\beta = \left\{ \left(\frac{1}{\sqrt{3}}, \frac{1}{\sqrt{3}}, \frac{1}{\sqrt{3}}\right), \left(\frac{1}{\sqrt{2}}, -\frac{1}{\sqrt{2}}, 0\right), \left(\frac{1}{\sqrt{6}}, \frac{1}{\sqrt{6}}, -\frac{2}{\sqrt{6}}\right) \right\}$

Verificar que cada vector tiene norma 1 y que son ortogonales entre sí (ejercicio).

Para encontrar las coordenadas de $v = (2, -1, 3)$:
$$a_1 = \langle v , v_1\rangle = (2)(1/\sqrt{3}) + (-1)(1/\sqrt{3}) + (3)(1/\sqrt{3}) = \frac{4}{\sqrt{3}}$$
$$a_2 = \langle v , v_2\rangle = (2)(1/\sqrt{2}) + (-1)(-1/\sqrt{2}) + (3)(0) = \frac{3}{\sqrt{2}}$$
$$a_3 = \langle v , v_3\rangle = (2)(1/\sqrt{6}) + (-1)(1/\sqrt{6}) + (3)(-2/\sqrt{6}) = \frac{2 - 1 - 6}{\sqrt{6}} = \frac{-5}{\sqrt{6}}$$

Por lo tanto $$[v]_\beta=\begin{pmatrix}\frac{4}{\sqrt{3}}\\\frac{3}{\sqrt{2}}\\\frac{-5}{\sqrt{6}}\end{pmatrix}$$

## Matriz de cambio de base

Un mismo vector $v$ tiene diferentes coordenadas según la base que elijamos. La **matriz de cambio de base** nos permite convertir coordenadas de una base a otra.

**Definición:** Sean $\beta = \{u_1, u_2, \dots, u_n\}$ y $\gamma = \{w_1, w_2, \dots, w_n\}$ dos bases de $\mathbb{R}^n$. La **matriz de cambio de base de $\beta$ a $\gamma$**, denotada $P_{\beta \to \gamma}$, es la matriz cuyas columnas son las coordenadas de los vectores de $\beta$ expresados en la base $\gamma$:
$$P_{\beta \to \gamma} = \begin{pmatrix} [u_1]_{\gamma} & [u_2]_{\gamma} & \cdots & [u_n]_{\gamma} \end{pmatrix}$$

**Propiedad:** Para cualquier vector $v \in V$:
$$[v]_{\gamma} = P_{\beta \to \gamma} \, [v]_{\beta}$$

**Matriz inversa:** La matriz de cambio de base inversa cumple:
$$P_{\gamma \to \beta} = P_{\beta \to \gamma}^{-1}$$

**Caso especial:** Cuando $\gamma$ es la base canónica, entonces $P_{\beta \to \text{can}}$ es simplemente la matriz cuyas columnas son los vectores de $\beta$:
$$P_{\beta \to \text{can}} = \begin{pmatrix} u_1 & u_2 & \cdots & u_n \end{pmatrix}$$

En este caso, $[v]_{\text{can}} = P_{\beta \to \text{can}} [v]_{\beta}$.

**Ejemplo 6. Cambio de base en $\mathbb{R}^2$**

Sean $\beta = \{(1,1), (1,-1)\}$ y $\gamma$ la base canónica $\{(1,0), (0,1)\}$.

Encontremos $P_{\beta \to \gamma}$:
$$[u_1]_{\gamma} = \begin{pmatrix} 1 \\ 1 \end{pmatrix}, \quad [u_2]_{\gamma} = \begin{pmatrix} 1 \\ -1 \end{pmatrix}$$
$$P_{\beta \to \gamma} = \begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}$$

Ahora, si $v = (2,3)$ en coordenadas canónicas, podemos encontrar $[v]_{\beta}$:
$$[v]_{\beta} = P_{\beta \to \gamma}^{-1} [v]_{\gamma}$$

Calculemos la inversa:
$$P_{\beta \to \gamma}^{-1} = -\frac{1}{2} \begin{pmatrix} -1 & -1 \\ -1 & 1 \end{pmatrix} = \begin{pmatrix} 0.5 & 0.5 \\ 0.5 & -0.5 \end{pmatrix}$$

Entonces:
$$[v]_{\beta} = \begin{pmatrix} 0.5 & 0.5 \\ 0.5 & -0.5 \end{pmatrix} \begin{pmatrix} 2 \\ 3 \end{pmatrix} = \begin{pmatrix} 2.5 \\ -0.5 \end{pmatrix}$$

**Ejemplo 7. Cambio entre dos bases no canónicas**

Sean $\beta = \{(1,1), (1,-1)\}$ y $\delta = \{(2,0), (0,2)\}$.

Primero, expresamos los vectores de $\beta$ en la base $\delta$:

Para $u_1 = (1,1)$ en base $\delta$: buscamos $a,b$ tales que $(1,1) = a(2,0) + b(0,2) = (2a, 2b)$. Entonces $a = 0.5$, $b = 0.5$. Así $[u_1]_{\delta} = (0.5, 0.5)^T$.

Para $u_2 = (1,-1)$: $(1,-1) = a(2,0) + b(0,2) = (2a, 2b)$. Entonces $a = 0.5$, $b = -0.5$. Así $[u_2]_{\delta} = (0.5, -0.5)^T$.

Por lo tanto:
$$P_{\beta \to \delta} = \begin{pmatrix} 0.5 & 0.5 \\ 0.5 & -0.5 \end{pmatrix}$$
