Sea $A$ una matriz cuadrada de tamaño $n \times n$. Un **eigenvector** de $A$ es un vector no nulo $v \in \mathbb{R}^n$ tal que:

$$A v = \lambda v$$

donde $\lambda$ es un escalar llamado **eigenvalor** (valor propio) asociado a $v$.

**Interpretación geométrica:** Multiplicar $A$ por $v$ simplemente escala a $v$ por el factor $\lambda$. La dirección de $v$ permanece invariante.

**Ejemplo:** Sea $A = \begin{pmatrix} 2 & 1 \\ 1 & 2 \end{pmatrix}$. Verifiquemos que $v = (1, 1)$ es eigenvector:

$$A v = \begin{pmatrix} 2 & 1 \\ 1 & 2 \end{pmatrix} \begin{pmatrix} 1 \\ 1 \end{pmatrix} = \begin{pmatrix} 3 \\ 3 \end{pmatrix} = 3 \begin{pmatrix} 1 \\ 1 \end{pmatrix}$$

Por lo tanto $\lambda = 3$ es el eigenvalor asociado.

## Cálculo de eigenvalores

La ecuación $A v = \lambda v$ puede reescribirse como:

$$(A - \lambda I) v = 0$$

Como $v \neq 0$, la matriz $(A - \lambda I)$ debe ser no invertible, es decir:

$$\det(A - \lambda I) = 0$$

Esta ecuación se llama **ecuación característica** y su lado izquierdo es el **polinomio característico** de $A$.

**Ejemplo:** Para $A = \begin{pmatrix} 2 & 1 \\ 1 & 2 \end{pmatrix}$:

$$A - \lambda I = \begin{pmatrix} 2-\lambda & 1 \\ 1 & 2-\lambda \end{pmatrix}$$

$$\det(A - \lambda I) = (2-\lambda)^2 - 1 = \lambda^2 - 4\lambda + 3 = (\lambda - 3)(\lambda - 1)$$

Por lo tanto los eigenvalores son $\lambda_1 = 3$ y $\lambda_2 = 1$.

```python
import numpy as np

A = np.array([[2, 1],
              [1, 2]])

# Cálculo directo de eigenvalores y eigenvectores
eigenvals, eigenvecs = np.linalg.eig(A)
```

## Propiedades de eigenvalores y eigenvectores

**Propiedades generales:**

1. **Traza:** $\text{tr}(A) = \lambda_1 + \lambda_2 + \dots + \lambda_n$
2. **Determinante:** $\det(A) = \lambda_1 \lambda_2 \cdots \lambda_n$
3. **Polinomio característico:** $p(\lambda) = \det(A - \lambda I)$
4. **Multiplicidad algebraica:** La multiplicidad de un eigenvalor como raíz del polinomio característico
5. **Multiplicidad geométrica:** La dimensión del espacio propio asociado (núcleo de $A - \lambda I$)

**Propiedades importantes:**
- La multiplicidad geométrica nunca excede la multiplicidad algebraica
- Si la multiplicidad geométrica es menor que la algebraica, la matriz es **defectuosa** (no diagonalizable)
- Eigenvectores asociados a eigenvalores distintos son linealmente independientes

## Diagonalización

Una matriz $A \in \mathbb{R}^{n \times n}$ es **diagonalizable** si existe una matriz invertible $P$ y una matriz diagonal $D$ tales que:

$$A = P D P^{-1}$$

En este caso, las columnas de $P$ son eigenvectores de $A$ y los elementos diagonales de $D$ son los eigenvalores correspondientes.

**Condición necesaria y suficiente:** $A$ es diagonalizable si y solo si tiene $n$ eigenvectores linealmente independientes (es decir, la multiplicidad geométrica de cada eigenvalor coincide con su multiplicidad algebraica).

**Diagonalización ortogonal (matrices simétricas):** Si $A$ es simétrica, entonces existe una matriz ortogonal $Q$ tal que:

$$A = Q \Lambda Q^\top$$

donde $\Lambda$ es diagonal con los eigenvalores y las columnas de $Q$ son eigenvectores ortonormales.

## Descomposición espectral

Para matrices simétricas $A$, la descomposición espectral es una forma especial de diagonalización:

$$A = \sum_{i=1}^n \lambda_i v_i v_i^\top$$

donde $\lambda_i$ son los eigenvalores y $v_i$ son los eigenvectores ortonormales.

En forma matricial: $A = V \Lambda V^\top$, donde $V$ es ortogonal.

**Interpretación:** Cada $v_i v_i^\top$ es una matriz de proyección ortogonal sobre el espacio generado por $v_i$.

**Aplicación en PCA:** En análisis de componentes principales, la descomposición espectral de la matriz de covarianza revela las direcciones de máxima varianza.
